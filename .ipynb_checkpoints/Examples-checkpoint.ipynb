### Pacce

This is a series of examples of different utilities you may find useful to use when measuring stellar population indices as well as comparing models and observations.
Pacce is a python code based upon standard python libraries such as numpy, scipy and (to use the wrapper) pandas.

```python
from pacce_wapper import pacce
```

We implemented a wrapper to the code in order to improve the user experience by letting the code handle the different stpes involved in the spectra preparations and measuremets.
You need the following:
- table contaning the name of spectra you want to measure. this can be as simple as a direct list of all the file names or a more complex table containing relevant infomration for you (e.g. redshifts, magnitudes)
- a directory containing the files you mention in the table
- an index definition file where you mention all the indices you would like to measure in youtr spectra. You can find an example of such a file in the support_files directory. Yo need 4 thing in each definition: a name for the index, a range where the index will be measured, continuum bands to anchor the continuum fit and a refentece to where you got thtat definition from.
- an output file to write the table containng the measurements you got.

An optional keyword that makes the code cleaner is the print_log option, where you can redirect the prints from the programs so you can take a look at what was performed in each spectra.

THe code returns a pandas data frame containing all the relevant info. Lets give it a try

```python
a = pacce(filename='./test_table.dat', path_to_files='./test_spectra/',
         IndexDefs='./suport_files/finalIndexDef.ind',
         output_file='indicies_MILES.txt', print_log='file.txt')
```

The code however can do much more that this, in the next example lets test the same structured but adding two additional keyword:
-  FWHM_ini is the fwhm intrinsic to the spectra you are feeding the code. This could also be R=lambda/d_lambda or sigma in km/s.
- sigma_fin is the sigma (km/s) that you want the final spectra to be measured in. As you can see, you can give two different keywords depending on the application (R_ini, sigma_fin, sigma_ini, FWHM_fin).

If you provide both of this info to the code (_ini and _fin), the code will perform the convolution of the spectra before measuring the indices.

A nice feature of the code is that you can add this info as a line in the table to be fed to the code i.e. a columns named 'sigma' will be interpreted as the 'sigma_ini' of that spectra. This can be useful if you hav data with different resolution and want them all to be measured in the same resolution. In this case you would just provide a _fin value and done!

Implementing this addition we have (notice in hte table there is a column names FWHM, so the code takes this info to use it as the _ini value):

```python
b = pacce(filename='./test_table.dat', path_to_files='./test_spectra/',
         IndexDefs='./suport_files/Elham_pacce_Defs.ind',
         output_file='indicies_MILES.txt', sigma_fin=300,print_log='file.txt')
```

You can also convert the indices you want to mag (as the standard output for the code is in $\AA$) by giving the index names you would like to convert. Also you can directly add new indices that are a combination of other indices (e.g. MgFe'). Use the columns names and add the expresion to a new text file where the code can read each line and feed it to the pandas.eval method (for troubleshooting and proper syntax, refer to pandas page)

```python
c = pacce(filename='./test_table.dat', path_to_files='./test_spectra/',
         IndexDefs='./suport_files/Elham_pacce_Defs.ind',
         output_file='indicies_MILES.txt', sigma_fin=300, A_to_mag=['Mg2', 'Mg1'], compute_idx='compute.txt', print_log='file.txt')
```

Pacce can also deal with uncertainties in the spectra i.e. if you provide an error spectrum the the code will provide errors for the measurements. This can be done in two ways two ways:
- the code estimates the error by calculating the S/N in the absorption and using the equations in https://arxiv.org/pdf/astro-ph/0606341.pdf to estimate the error. this is a faster approach but does not account for the variation in the continuum. If you provide the error spectra, this option will be automatically triggered.
- the code estimates the error using an MC approach i.e. using the error provided, generated mock specra based upon the oberved one. The EW is measured several times and the standard deviation is used as the error. This is a more robust approach, but also more time consuming. In order to trigger this oprtion you need to provide the number of iterations to perform this estimation i.e. simulate=50.

```python
c = pacce(filename='./test_table.dat', path_to_files='./test_spectra_fake_err/',IndexDefs='./suport_files/Elham_pacce_Defs.ind',simulate=50, output_file='indicies_MILES_sim.txt',
          compute_idx='compute.txt', print_log='file.txt')
```

```python
d = pacce(filename='./test_table.dat', path_to_files='./test_spectra_fake_err/',IndexDefs='./suport_files/Elham_pacce_Defs.ind', output_file='indicies_MILES.txt',
          compute_idx='compute.txt', print_log='file.txt', FWHM_fin=5.0)
```

You can also have a look at the changes implemented by the code as well as the continuum fit and location of the bands. You just need to provide  a directory where you can save the plots from the code and the code will automatically plot all the indices. This should also be used mindful that it takes a lot more time to complete this operation.

```python
e = pacce(filename='./test_table.dat', path_to_files='./test_spectra_fake_err/',IndexDefs='./suport_files/Elham_pacce_Defs.ind', output_file='indicies_MILES.txt',
          compute_idx='compute.txt', print_log='file.txt', path_plots='./Pacce_figures/', FWHM_fin=5.0)
```
